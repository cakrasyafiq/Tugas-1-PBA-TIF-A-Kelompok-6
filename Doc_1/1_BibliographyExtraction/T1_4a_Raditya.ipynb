```python
# Dataset Mentah (Source Text)
RAW_DOC1_DATA = """Goldberg, Yoav (2016). "A Primer on Neural Network Models for Natural Language Processing". Journal of Artificial Intelligence Research. 57: 345–420. arXiv:1807.10854.

Goodfellow, Ian; Bengio, Yoshua; Courville, Aaron (2016). Deep Learning. MIT Press.

Jozefowicz, Rafal; Vinyals, Oriol; Schuster, Mike; Shazeer, Noam; Wu, Yonghui (2016). Exploring the Limits of Language Modeling. arXiv:1602.02410.

Choe, Do Kook; Charniak, Eugene. "Parsing as Language Modeling". EMNLP 2016.

Vinyals, Oriol; et al. "Grammar as a Foreign Language" (PDF). NIPS2015.

Winograd, Terry (1971). Procedures as a Representation for Data in a Computer Program for Understanding Natural Language (Thesis).

Schank, Roger C.; Abelson, Robert P. (1977). Scripts, Plans, Goals, and Understanding: An Inquiry Into Human Knowledge Structures. Hillsdale: Erlbaum. ISBN 0-470-99033-3.

Mark Johnson. How the statistical revolution changes (computational) linguistics. Proceedings of the EACL 2009 Workshop on the Interaction between Linguistics and Computational Linguistics.

Philip Resnik. Four revolutions. Language Log, February 5, 2011.

Klein, Dan; Manning, Christopher D. (2002). "Natural language grammar induction using a constituent-context model" (PDF). Advances in neural information processing systems.

Kishorjit, N.; Vidya, Raj RK.; Nirmal, Y.; Sivaji, B. (2012). "Manipuri Morpheme Identification" (PDF). Proceedings of the 3rd Workshop on South and Southeast Asian Natural Language Processing (SANLP). COLING 2012, Mumbai, December 2012: 95–108.

Duan, Yucong; Cruz, Christophe (2011). "Formalizing Semantic of Natural Language through Conceptualization from Existence". International Journal of Innovation, Management and Technology. 2 (1): 37–42. Archived from the original on 2011-10-09.

Mittal (2011). "Versatile question answering systems: seeing in synthesis". IJIIDS. 5 (2): 119–142. doi:10.1504/IJIIDS.2011.038968.

PASCAL Recognizing Textual Entailment Challenge (RTE-7) https://tac.nist.gov//2011/RTE/

Yi, Chucai; Tian, Yingli (2012), "Assistive Text Reading from Complex Background for Blind Persons", Camera-Based Document Analysis and Recognition, Springer Berlin Heidelberg, pp. 15–28, CiteSeerX 10.1.1.668.869, doi:10.1007/978-3-642-29364-1_2, ISBN 9783642293634."""


def create_input_file(filename):
    """Membaca/membuat file input doc_1.txt jika belum tersedia."""
    if not os.path.exists(filename):
        with open(filename, "w", encoding="utf-8") as f:
            f.write(RAW_DOC1_DATA)
        print(f"[INFO] File masukan '{filename}' berhasil dibuat.")


def load_input_text(filename):
    """Membaca isi file masukan dan mengembalikan teks utuh."""
    create_input_file(filename)
    with open(filename, "r", encoding="utf-8") as f:
        return f.read()
```

# Pre-Processing Teks & Pemisahan Entri

```python
def preprocess_entries(content):
    """Memisah isi berkas berdasarkan baris kosong dan membersihkan newline berlebih."""
    raw_entries = [e.strip() for e in content.split("\n\n") if e.strip()]
    cleaned_entries = [re.sub(r"\s+", " ", entry) for entry in raw_entries]
    return cleaned_entries
```

# Ekstraksi Informasi dengan Regular Expression (Regex)

```python
def extract_single_entry(entry_text):
    """Mengekstrak atribut authors, year, dan title dari satu baris entri referensi."""
    # Pola Regex untuk mencocokkan pola bibliografi
    pattern = re.compile(
        r"^(?P<authors>.+?)\s*(?:\((?P<year>\d{4})\)\.?)?\s*(?:\"(?P<title_quoted>[^\"]+)\"|(?P<title_unquoted>[A-Z][^.\n]+))\."
    )

    match = pattern.search(entry_text)
    if not match:
        return None

    authors = match.group("authors").strip()
    year = match.group("year")
    title = match.group("title_quoted") or match.group("title_unquoted")

    entry_dict = {"authors": authors}

    # Atribut 'year' hanya ditambahkan jika ditemukan (sesuai spesifikasi soal)
    if year:
        entry_dict["year"] = year

    if title:
        entry_dict["title"] = title.strip()

    return entry_dict
```

# Penyimpanan Hasil ke JSON Output

```python
def save_to_json(data_list, output_filename):
    """Menyimpan list dictionary hasil ekstraksi ke file JSON dengan format rapi."""
    with open(output_filename, "w", encoding="utf-8") as f:
        json.dump(data_list, f, indent=4, ensure_ascii=False)
    print(
        f"[BERHASIL] Output berhasil disimpan di '{output_filename}' ({len(data_list)} entri)."
    )
```

# Main

```python
def main():
    input_file = "doc_1.txt"
    output_file = "a_judul.json"

    # 1. Load teks input
    content = load_input_text(input_file)

    # 2. Preprocessing & pisah entri
    entries = preprocess_entries(content)

    # 3. Ekstraksi tiap entri dengan Regex
    extracted_data = []
    for entry in entries:
        parsed_dict = extract_single_entry(entry)
        if parsed_dict:
            extracted_data.append(parsed_dict)

    # 4. Simpan ke JSON
    save_to_json(extracted_data, output_file)


if __name__ == "__main__":
    main()
```
